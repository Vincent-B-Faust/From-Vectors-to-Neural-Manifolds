# From Vectors to Neural Manifolds: Geometric Principles of Brain Computation

## English Version Lecture Notes

**Target Audience:** Neuroscience students without computational backgrounds  
**Learning Path:** Vectors → Linear Transformations → Frequency Domain (Fourier/Wavelets) → Sparsity → Neural Manifolds

---

### 🎯 Core Logic Chain

- **Representation**: How neural activity translates into points and trajectories in vector space
- **Transformation**: How neural networks process information through matrix operations and basis transformations
- **Structure**: How this processing reveals low-dimensional geometric structures in high-dimensional data

---

### 📚 Learning Objectives

Upon completing this lecture, you will be able to:

1. Express population neural activity as $\mathbf{r}(t)\in\mathbb{R}^N$ and reason about "neural state space trajectories"
2. Use dot products and cosine similarity to interpret "neural matching," "template responses," and "decorrelation"
3. Use matrices to explain inter-region projections and understand determinants and eigenvalues
4. Understand Fourier transforms as coordinate system transformations
5. Handle non-stationary neural signals with STFT and wavelets
6. Understand why sparsity implies low-dimensional structure
7. Use PCA/Isomap to reveal hidden manifold structures
8. Understand Takens time-delay embedding and dynamic systems
